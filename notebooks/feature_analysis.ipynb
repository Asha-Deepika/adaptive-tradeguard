```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load feature matrix
df = pd.read_csv("../data/processed/feature_matrix.csv")

# Convert Date column to datetime
df["Date"] = pd.to_datetime(df["Date"])

print("Data loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["Close"])

plt.title("NIFTY 50 Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Closing Price")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["Return"])

plt.title("Daily Return Over Time")
plt.xlabel("Date")
plt.ylabel("Return")

plt.axhline(0, linestyle="--")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["20-day Volatility"])

plt.title("20-Day Volatility Over Time")
plt.xlabel("Date")
plt.ylabel("Volatility")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["RSI-14"])

plt.axhline(70, linestyle="--", label="Overbought (70)")
plt.axhline(30, linestyle="--", label="Oversold (30)")

plt.title("RSI-14 Over Time")
plt.xlabel("Date")
plt.ylabel("RSI")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["MACD"], label="MACD")
plt.plot(df["Date"], df["MACD Signal"], label="Signal")

plt.axhline(0, linestyle="--")

plt.title("MACD and Signal Line")
plt.xlabel("Date")
plt.ylabel("MACD")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.bar(
    df["Date"],
    df["MACD Histogram"]
)

plt.axhline(0, linestyle="--")

plt.title("MACD Histogram")

plt.xlabel("Date")
plt.ylabel("MACD Histogram")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df["Date"], df["ATR-14"])

plt.title("ATR-14 Over Time")
plt.xlabel("Date")
plt.ylabel("ATR-14")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Date"],
    df["Volume"],
    label="Volume"
)

plt.plot(
    df["Date"],
    df["Volume MA-20"],
    label="Volume MA-20"
)

plt.title(
    "Trading Volume and 20-Day Volume Moving Average"
)

plt.xlabel("Date")
plt.ylabel("Volume")

plt.legend()
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(
    df["Date"],
    df["Volume Change"]
)

plt.axhline(0, linestyle="--")

plt.title("Volume Change Over Time")

plt.xlabel("Date")
plt.ylabel("Volume Change")

plt.grid(True)
plt.tight_layout()

plt.show()
```

```python
import seaborn as sns

feature_columns = [
    "Return",
    "Log Return",
    "20-day Volatility",
    "RSI-14",
    "MACD",
    "MACD Signal",
    "MACD Histogram",
    "ATR-14",
    "Volume Change",
    "Volume MA-20"
]

correlation = df[feature_columns].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Heatmap")
plt.tight_layout()

plt.show()
```
